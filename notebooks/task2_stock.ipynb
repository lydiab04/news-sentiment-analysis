```python
import pandas as pd
import matplotlib.pyplot as plt
stock = pd.read_csv("../data/raw/yfinance_data/Data/AAPL.csv")
stock.head()
```

```python
stock["Date"] = pd.to_datetime(stock["Date"])
stock = stock.sort_values("Date")
stock = stock.dropna()
```

```python
stock["SMA_20"] = stock["Close"].rolling(20).mean()
```

```python
plt.figure(figsize=(12,6))
plt.plot(stock["Date"], stock["Close"], label="Close")
plt.plot(stock["Date"], stock["SMA_20"], label="SMA 20")
plt.title("Closing Price with Simple Moving Average")
plt.legend()
plt.show()
```
