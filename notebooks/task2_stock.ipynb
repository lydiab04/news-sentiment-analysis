```python
import pandas as pd
import matplotlib.pyplot as plt
stock = pd.read_csv("../data/raw/yfinance_data/Data/AAPL.csv")
stock.head()
```

```python
import talib

stock["RSI"] = talib.RSI(stock["Close"], timeperiod=14)

macd, signal, hist = talib.MACD(stock["Close"])

stock["MACD"] = macd
stock["MACD_SIGNAL"] = signal
```

```python
stock["Date"] = pd.to_datetime(stock["Date"])
stock = stock.sort_values("Date")
stock = stock.dropna()
```

```python
plt.figure(figsize=(12,5))
plt.plot(stock["Date"], stock["RSI"])
plt.title("RSI Indicator")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(stock["Date"], stock["MACD"], label="MACD")
plt.plot(stock["Date"], stock["MACD_SIGNAL"], label="Signal")
plt.legend()
plt.title("MACD Indicator")
plt.show()
```

```python
stock["SMA_20"] = stock["Close"].rolling(20).mean()
```

```python
plt.figure(figsize=(12,6))
plt.plot(stock["Date"], stock["Close"], label="Close")
plt.plot(stock["Date"], stock["SMA_20"], label="SMA 20")
plt.title("Closing Price with Simple Moving Average")
plt.legend()
plt.show()
```
